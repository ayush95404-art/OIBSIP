```python
import pandas as pd
import numpy as np

# Load Titanic dataset
df = pd.read_csv("https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv")

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

df.head()
```

```python
# TASK 3 - DATA QUALITY REPORT

print("DATA QUALITY REPORT")
print("=" * 50)

# 1. Missing values
print("\n1. Missing Values:")
print(df.isnull().sum())

# 2. Duplicate rows
print("\n2. Duplicate Rows:")
print(df.duplicated().sum())

# 3. Data types
print("\n3. Data Types:")
print(df.dtypes)

# 4. Numerical summary and range anomalies
print("\n4. Numerical Summary:")
display(df.describe())

# 5. Unique values
print("\n5. Unique Values:")
print(df.nunique())

print("\nData quality report completed successfully!")
```

```python
# TASK 3 - MISSING DATA HANDLING

# Fill Age with median
df["Age"] = df["Age"].fillna(df["Age"].median())

# Fill Embarked with mode
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])

# Drop Cabin because it has too many missing values
df = df.drop(columns=["Cabin"])

print("Missing data handled successfully!")
print("\nRemaining missing values:")
print(df.isnull().sum())
```

```python
# TASK 3 - DUPLICATE REMOVAL

duplicates_before = df.duplicated().sum()

print("Duplicate rows before cleaning:", duplicates_before)

df = df.drop_duplicates()

print("Duplicate rows after cleaning:", df.duplicated().sum())
print("Rows remaining:", len(df))
```

```python
# TASK 3 - STANDARDISATION

# Standardise Sex values
df["Sex"] = df["Sex"].str.lower().str.strip()

# Standardise Embarked values
df["Embarked"] = df["Embarked"].str.upper().str.strip()

print("Standardisation completed successfully!")

print("\nSex values:")
print(df["Sex"].unique())

print("\nEmbarked values:")
print(df["Embarked"].unique())
```

```python
# TASK 3 - OUTLIER DETECTION

import numpy as np

numeric_columns = ["Age", "Fare", "SibSp", "Parch"]

for col in numeric_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = ((df[col] < lower) | (df[col] > upper)).sum()

    print(f"{col}: {outliers} outliers")

print("\nOutlier detection completed.")
```

```python
# TASK 3 - OUTLIER TREATMENT USING IQR CAPPING

for col in ["Age", "Fare", "SibSp", "Parch"]:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    # Cap values at the IQR limits
    df[col] = df[col].clip(lower=lower, upper=upper)

print("Outlier treatment completed successfully!")

print("\nUpdated numerical summary:")
display(df[["Age", "Fare", "SibSp", "Parch"]].describe())
```

```python
# TASK 3 - FINAL DATA QUALITY CHECK

print("FINAL DATA QUALITY REPORT")
print("=" * 50)

print("\nShape of cleaned dataset:")
print(df.shape)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nData types:")
print(df.dtypes)

print("\nFirst 5 cleaned rows:")
display(df.head())

print("\nData cleaning completed successfully!")
```

```python
# TASK 3 - SAVE CLEANED DATASET

df.to_csv("Titanic_Cleaned.csv", index=False)

print("Cleaned dataset saved successfully as Titanic_Cleaned.csv")
```
